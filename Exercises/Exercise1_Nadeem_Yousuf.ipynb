```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
# Total values to be generated
num_values=1000

# Generate f1,f2,f3,f4 randomly
f1 = np.random.uniform(18,75,num_values)
f2 = np.random.randint(1,18,num_values)
f3 = np.random.binomial(10,0.2,num_values)
f4 = np.random.normal(50,10,num_values)
```

```python
my_df = pd.DataFrame(
    {
        'f1':f1,
        'f2':f2,
        'f3':f3,
        'f4':f4
    }
)
my_df.head(10)
```

```python
# create a scatterplot from f2 and f3 using seaborn
sns.scatterplot(
    data=my_df,
    x='f2',
    y='f3',
    color='blue',
    s=80
    
)
plt.title('Relationship between f1 and f2')
plt.xlabel("f2 values")
plt.ylabel("f3 values")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# calculate mean rowwise and store that in f5
my_df['f5'] = my_df.mean(axis=1)
my_df.head(10)
```

```python
#export to excel
my_df.to_excel('exercise1.xlsx',index=False)
```
