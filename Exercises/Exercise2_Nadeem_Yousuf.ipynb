```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_cat=cv2.imread(r'cat.jpg')
image_dog = cv2.imread(r'dog.jpg')
print(f'Size of cat image {image_cat.shape}')
print(f'Size of dog image {image_dog.shape}')
```

```python
if image_cat.shape != image_dog.shape:
    image_cat = cv2.resize(image_cat, (image_dog.shape[1], image_dog.shape[0]))
```

```python
image_blended=cv2.addWeighted(image_cat,0.5,image_dog,0.5,0)
cv2.imwrite('image_blended.jpg', image_blended)
```

```python
# 5. Convert BGR (OpenCV) to RGB (Matplotlib) for accurate colors
image_cat_rgb = cv2.cvtColor(image_cat, cv2.COLOR_BGR2RGB)
image_dog_rgb = cv2.cvtColor(image_dog, cv2.COLOR_BGR2RGB)
image_blended_rgb = cv2.cvtColor(image_blended, cv2.COLOR_BGR2RGB)

# 6. Display side-by-side inline plots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

images = [image_cat_rgb, image_dog_rgb, image_blended_rgb]
titles = ['Input Image 1', 'Input Image 2', 'Blended Output']

for ax, img, title in zip(axes, images, titles):
    ax.imshow(img)
    ax.set_title(title)
    ax.axis('off')  # Hide pixel coordinate axes

plt.tight_layout()
plt.show()
```
